```python
import pandas as pd

df = pd.read_csv("Global Superstore.csv")
print(df.shape)
print(df.columns.tolist())
```

```python
print(df.shape)
print(df.columns.tolist())
```

```python
print("Shape:", df.shape)
print("\nColumn types:")
print(df.dtypes)
print("\nFirst 3 rows:")
print(df.head(3))
```

```python
print(df[['Sales', 'Profit', 'Quantity', 'Discount']].describe())
```

```python
# Missing values
print("Missing values per column:")
print(df.isnull().sum()[df.isnull().sum() > 0])

# Duplicate rows
print("\nDuplicate rows:", df.duplicated().sum())

# Negative quantity (invalid data असेल तर)
print("\nNegative Quantity entries:", (df['Quantity'] < 0).sum())

# Outliers in Sales (IQR method)
Q1 = df['Sales'].quantile(0.25)
Q3 = df['Sales'].quantile(0.75)
IQR = Q3 - Q1
outliers = df[(df['Sales'] < Q1 - 1.5*IQR) | (df['Sales'] > Q3 + 1.5*IQR)]
print("\nSales outliers (IQR method):", len(outliers))
```

```python
print("Average Sales by Category:")
print(df.groupby('Category')['Sales'].mean().sort_values(ascending=False))

print("\nTotal Profit by Region:")
print(df.groupby('Region')['Profit'].sum().sort_values(ascending=False))

print("\nYearly Sales Trend:")
print(df.groupby('Year')['Sales'].sum())
```

```python
from scipy import stats

# High discount vs Low discount groups मध्ये Profit वेगळं आहे का?
median_discount = df['Discount'].median()
high_discount = df[df['Discount'] > median_discount]['Profit']
low_discount = df[df['Discount'] <= median_discount]['Profit']

t_stat, p_val = stats.ttest_ind(high_discount.dropna(), low_discount.dropna())
print(f"T-statistic: {t_stat:.3f}")
print(f"P-value: {p_val:.5f}")

if p_val < 0.05:
    print("=> Statistically significant! Discount level significantly affects Profit.")
else:
    print("=> No statistically significant difference.")
```
